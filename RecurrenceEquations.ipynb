# Recurrence equations

## Delayed logistic equation (Mathematical Biology sheet 2, question 3)

$$ x_{n+1} = r x_{n} (1 - x_{n-1}) $$

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
r = 2.5
nmax = 20
x = np.empty(nmax)  # full logistic equation
eps = np.empty(nmax)  # linearised equation

xeq = 1 - r**-1

x[0] = xeq + 0.01
x[1] = xeq + 0.01
eps[0] = 0.01
eps[1] = 0.01
for n in range(2, nmax):
    x[n] = r * x[n-1] * (1 - x[n-2])
    eps[n] = eps[n-1] + (1-r) * eps[n-2]
```

```python
ax = plt.gca()
ax.plot(x, 'r-', label=r'$x_n$')
ax.plot(xeq + eps, 'b--', label=r'$x^* + \epsilon_n$')
ax.hlines(xeq, 0, nmax, colors='black', linestyles='--', label='$x^*$')
ax.legend()
# ax.set_ylim(-0.05, 1.5)
```
