# Laplace’s equation

This demo uses the `py-pde` and `napari` packages to solve Laplace’s
equation in a few different domains.

```python
import numpy as np
import matplotlib.pyplot as plt
import pde
from pde import CartesianGrid, solve_laplace_equation
```

```python
grid = pde.UnitGrid([64, 64])                 # generate grid
state = pde.ScalarField.random_uniform(grid)  # generate initial condition

eq = pde.DiffusionPDE(diffusivity=0.1)        # define the pde
result = eq.solve(state, t_range=10)          # solve the pde
result.plot()                                 # plot the resulting field
```

```python
grid = CartesianGrid([[0, 2 * np.pi]] * 2, 64)
bcs = [{"value": "y"}, {"value": "sin(x)"}]

res = solve_laplace_equation(grid, bcs)
pl = res.plot()
```

```python
from pde import CylindricalGrid, ScalarField

grid = CylindricalGrid(radius=3, bounds_z=[0, 4], shape=16)
field = ScalarField.from_expression(grid, "sqrt(z) * exp(-r**2)")
field.plot(title="Scalar field in cylindrical coordinates")
```
